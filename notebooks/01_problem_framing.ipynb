# Home Credit Default Risk - Problem Framing

## 1. Business Context
- Home Credit is a Non-banking financial company (NBFC) and they provide consumer loans such as:
  - Mobile phones
  - Home appliances
  - Personal loans
- These Loans are offered with minimal documents, means:
  - Higher accessibility
  - Higher credit risk exposure

Because many customers may have limited traditional credit history, the company must rely on alternative data-driven risk modeling.

## 2. Why Credit Risk Matters

Credit risk determines whether the lender will lose money due to borrower default.

If a risky customer is approved:
- Principal may not be recovered
- Expected interest income is lost
- Recovery or legal costs increase
- Regulatory capital burden increases

These directly impact profitability.

## 3. Problem Definition
This is a **Binary classification problem.**

### Target variable
The dataset contains a target column named `TARGET`, defined as:
```bash
  • Target = 1 → Client will default (difficulty in recovery)
  • Target = 0 → Client will repay
```
In this context, the **positive class is Default (1)**, as it represents the risk event the business aims to predict.

### Modeling Objective

The goal is to estimate:

$$
P(\text{Default} \mid \text{Customer Features})
$$

## 4. Expected Loss Framework
Expected Loss (EL) is defined as:

    EL=PD×LGD×EAD

Where:
- PD → Probability of Default (model output)
- LGD → Loss Given Default
- EAD → Exposure At Default (loan amount)

### Business trade-off:
| Strategy | Outcome |
|----------|---------|
| Approve everyone | High defaults |
| Reject everyone |	No revenue |
| Smart risk-based approval |	Optimized profit |

Therefore, the model must rank customers by risk, not just classify them.

## 5. Metric Justification (ROC-AUC)
The dataset is imbalanced (~8% default rate), Accuracy would be misleading.

Example:

If model predicts all as non-default:
- Accuracy ≈ 92%
- But zero usefulness

We use:

ROC-AUC (Receiver Operating Characteristic – Area Under Curve)

Because:

- It measures ranking ability

- It is threshold-independent

- It works well with imbalanced data

ROC-AUC evaluates how well the model separates defaulters from non-defaulters.

## 6. ML Translation
### Input:
- Customer demographics
- Financial information
- Behavioral history
- Credit bureau data

### Output:
- Probability score between 0 and 1

Instead of Predict 0 or 1, We think in terms of Assigning a risk score.


Example:

| Risk Score | Decision |
|------------|-----------|
| 0–0.2 | Approve |
| 0.2–0.5 | Review |
| >0.5 | Reject |
